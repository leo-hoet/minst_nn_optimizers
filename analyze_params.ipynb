```python
import pandas as pd
import json
import seaborn as sns

file_name = 'fitnesses.json'
with open(file_name, 'r') as f:
    data = f.read()

parsed_data = json.loads(data)
runs_data = parsed_data['runs']

df_all = pd.DataFrame(runs_data)
df_all
```

```python
df = df_all[['w','c_local','c_global', 'pop_size']]
df['best'] = df_all['fitness'].apply(lambda x: x[-1])
df.describe()
```

```python
sns.pairplot(df, hue='best')
```
